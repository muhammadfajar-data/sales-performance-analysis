```python
# Sales Decline Analysis: 2015 vs 2016

sales_comparison = (
    df[df['Order Date'].dt.year.isin([2015, 2016])]
    .groupby([df['Order Date'].dt.year, 'Category'])['Sales']
    .sum()
    .unstack()
)

sales_comparison.loc['Change'] = (
    sales_comparison.loc[2016] - sales_comparison.loc[2015]
)

sales_comparison.loc['Change (%)'] = (
    (sales_comparison.loc[2016] - sales_comparison.loc[2015])
    / sales_comparison.loc[2015] * 100
)

sales_comparison.round(2)
```

```python
# Office Supplies: Product Sales Comparison (2015 vs 2016)

office_supplies = df[
    (df['Category'] == 'Office Supplies') &
    (df['Order Date'].dt.year.isin([2015, 2016]))
].copy()

office_supplies['Year'] = office_supplies['Order Date'].dt.year

product_sales = office_supplies.pivot_table(
    index='Product Name',
    columns='Year',
    values='Sales',
    aggfunc='sum',
    fill_value=0
)

# Pastikan kedua kolom tahun tersedia
for year in [2015, 2016]:
    if year not in product_sales.columns:
        product_sales[year] = 0

product_sales['Change'] = product_sales[2016] - product_sales[2015]

product_sales['Change (%)'] = (
    product_sales['Change'] /
    product_sales[2015].replace(0, float('nan')) * 100
)

product_sales.sort_values('Change').head(10).round(2)
```

```python
# Office Supplies: Product Performance Overview

office_sales = df[df['Category'] == 'Office Supplies'].copy()

office_sales['Year'] = office_sales['Order Date'].dt.year

product_yearly_sales = office_sales.pivot_table(
    index='Product Name',
    columns='Year',
    values='Sales',
    aggfunc='sum',
    fill_value=0
)

# Pastikan kedua tahun tersedia
for year in [2015, 2016]:
    if year not in product_yearly_sales.columns:
        product_yearly_sales[year] = 0

# Klasifikasi perubahan penjualan
product_yearly_sales['Status'] = 'No Sales Both Years'

product_yearly_sales.loc[
    (product_yearly_sales[2015] > 0) &
    (product_yearly_sales[2016] == 0),
    'Status'
] = 'No Sales in 2016'

product_yearly_sales.loc[
    (product_yearly_sales[2015] == 0) &
    (product_yearly_sales[2016] > 0),
    'Status'
] = 'New Sales in 2016'

product_yearly_sales.loc[
    product_yearly_sales[2016] > product_yearly_sales[2015],
    'Status'
] = 'Increased'

product_yearly_sales.loc[
    (product_yearly_sales[2016] > 0) &
    (product_yearly_sales[2016] < product_yearly_sales[2015]),
    'Status'
] = 'Decreased'

product_yearly_sales['Status'].value_counts()
```
